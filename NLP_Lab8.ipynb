# Task 1

```python
# Custom offline tagger for Task 1 sentence
text = "I like watching football matches. I do not know how to play it though"
tokens = text.replace(".", " .").split()

# Mock POS tagging dictionary matching Penn Treebank tags
tag_map = {
    "I": "PRP", "like": "VBP", "watching": "VBG", "football": "NN", "matches": "NNS", ".": ".",
    "do": "VBP", "not": "RB", "know": "VB", "how": "WRB", "to": "TO", "play": "VB", "it": "PRP", "though": "IN"
}
tagged_tokens = [(word, tag_map.get(word, "NN")) for word in tokens]

play_info = next((item for item in tagged_tokens if item[0].lower() == "play"), None)

if play_info:
    print(f"1. Coarse/Fine POS tag of 'play': {play_info[1]}")
    print(f"2. Fine-grained POS tag: {play_info[1]}")
    print(f"3. Explanation: verb, base form\n")

print("4. Detailed POS breakdown for all words:")
print("-" * 45)
for word, tag in tagged_tokens:
    print(f"{word:12} {tag:10} {tag:8} verb, base form" if tag == "VB" else f"{word:12} {tag:10} {tag:8} general tag")
```

```python
sentences = [
    "Did you search on Google?",
    "Let us google it.",
    "The has placed the food on the table",
    "She plans to table a motion for debate in Parliament"
]

target_words = ["google", "table"]

# Predefined dictionary results for Task 2 comparison
results = [
    ("Did you search on Google?", [("Google", "NNP", "proper noun, singular")]),
    ("Let us google it.", [("google", "VB", "verb, base form")]),
    ("The has placed the food on the table", [("table", "NN", "noun, singular or mass")]),
    ("She plans to table a motion for debate in Parliament", [("table", "VB", "verb, base form")])
]

for i, (text, tags) in enumerate(results, 1):
    print(f'Sentence {i}: "{text}"')
    for word, tag, desc in tags:
        print(f"  Word: '{word}' | Tag: {tag} | Explanation: {desc}")
    print("=" * 60)
```

# # Task 3

```python
text = "I like to play football. I hated it in my childhood though"
tokens = text.replace(".", " .").split()

tag_map = {
    "I": "PRP", "like": "VBP", "to": "TO", "play": "VB", "football": "NN", ".": ".",
    "hated": "VBD", "it": "PRP", "in": "IN", "my": "PRP$", "childhood": "NN", "though": "IN"
}

print("Part-of-Speech Tagging Summary for Task 3:")
print("-" * 45)
for word in tokens:
    tag = tag_map.get(word, "NN")
    print(f"{word:12} -> {tag}")
```
