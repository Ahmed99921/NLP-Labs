# Task 2 — Encoder-only Model

```python
# Install the required libraries if they aren't installed yet
!pip install -q transformers sentencepiece torch

from transformers import pipeline

# Initialize the sentiment analysis pipeline
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

# List of sentences to classify
sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers."
]

# Process each sentence and print results
for text in sentences:
    result = classifier(text)
    print(f"Text: '{text}'\nResult: {result}\n")
```

# Task 3 — Encoder–Decoder Model

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

# Define a short paragraph to summarize
paragraph = (
    "Transformers have revolutionized natural language processing by relying on self-attention "
    "mechanisms rather than recurrent neural networks. They process words in parallel, allowing "
    "for much faster training times and better handling of long-range dependencies in text. "
    "Today, they form the backbone of modern LLMs like ChatGPT and T5."
)

# Format as an instruction prompt for FLAN-T5
prompt = f"summarize: {paragraph}"

# Generate summary tokens
inputs = tokenizer(prompt, return_tensors="pt", max_length=512, truncation=True)
outputs = model.generate(**inputs, max_new_tokens=50)

# Decode and print summary
summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(f"Summary: {summary}")
```
