# Task 1: Cosine Similarity

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Define the sentences
sentences = [
    'This is the first document.',  
    'This document is the second document.',  
    'And this is the third one.',  
    'Is this the first document?'  
]

# Apply TF-IDF and calculate cosine similarity
tfidf_vectorizer = TfidfVectorizer()
vector_matrix = tfidf_vectorizer.fit_transform(sentences)
cosine_sim_matrix = cosine_similarity(vector_matrix)

# Create a DataFrame for clear visualization
df_cosine = pd.DataFrame(data=cosine_sim_matrix, index=sentences, columns=sentences)
df_cosine
```

# Task 2: TF-IDF

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Define the sentences
sentences = [
    'data science is one of the most important fields of science',  
    'this is one of the best data science courses',  
    'data scientists analyze data'  
]

# Apply TF-IDF vectorization
tfidf_vectorizer = TfidfVectorizer()
result = tfidf_vectorizer.fit_transform(sentences)

# Convert to DataFrame to see the important words and their TF-IDF scores
df_tfidf = pd.DataFrame(result.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
df_tfidf
```

# Task 3: Word2Vec (Simpsons Dataset Preprocessing & Model Training)

```python
import os
import re
import pandas as pd
import gensim
import kagglehub

# Automatically download/locate The Simpsons dataset using kagglehub
path = kagglehub.dataset_download("prashant111/the-simpsons-dataset")

# Search for the CSV file path in the downloaded folder
csv_path = None
for root, dirs, files in os.walk(path):
    if "simpsons_script_lines.csv" in files:
        csv_path = os.path.join(root, "simpsons_script_lines.csv")
        break

if not csv_path:
    raise FileNotFoundError("simpsons_script_lines.csv could not be found.")

# Load the Simpsons script lines dataset
df_simpsons = pd.read_csv(csv_path, engine='python', on_bad_lines='skip')

# Drop rows with missing values in 'spoken_words'
df_simpsons = df_simpsons.dropna(subset=['spoken_words'])

# Preprocessing function specified in the lab task
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

# Tokenize all spoken words using regular expressions
tokens = []
for text in df_simpsons['spoken_words']:
    cleaned = clean_text(text)
    token = re.findall(r'\b\w+\b', cleaned)
    tokens.append(token)

# Train the Skip-gram Word2Vec model
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1)
print("Skip-gram Word2Vec model trained successfully!")
```
