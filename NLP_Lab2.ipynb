# Task 1: Hashtag Extraction and Counting

```python
import pandas as pd
import re
from collections import Counter

# Load the dataset (replace 'apple_tweets.csv' with your actual downloaded file name/path)
# df = pd.read_csv('apple_tweets.csv')

# Example dataframe simulation for demonstration:
data = {'text': ['Loving the new #Apple event! #iPhone16', 
                 'Check out this amazing #MacbookPro #Apple', 
                 'So excited for #iOS18 and #Apple updates!']}
df = pd.DataFrame(data)

# Extract all hashtags using regular expressions (matches '#' followed by alphanumeric characters)
all_hashtags = []
for tweet in df['text']:
    hashtags = re.findall(r'#\w+', tweet)
    all_hashtags.extend(hashtags)

# 1. Find total number of hashtags
total_hashtags = len(all_hashtags)
print(f"Total number of hashtags: {total_hashtags}")

# 2. Find the top 10 hashtags
top_10_hashtags = Counter(all_hashtags).most_common(10)
print("\nTop 10 Hashtags:")
for tag, count in top_10_hashtags:
    print(f"{tag}: {count}")
```

# Task 2: Using re.compile()

```python
import re

text = "This year is 2021"

# 1. Use re.compile() to create a regular expression pattern that matches one or more digits.
pattern = re.compile(r"\d+")

# 2. Print the type of the compiled pattern.
print("Type of compiled pattern:", type(pattern))

# 3. Use the compiled pattern to replace 2021 with 2022.
updated_text = pattern.sub("2022", text)

# 4. Print the updated text.
print("Updated text:", updated_text)
```

# Task 3: Using re.split()

```python
import re

text = "a 11 b 2 3 c 4"

# 1. Use re.split() with a regular expression pattern to split the text whenever one or more digits appear.
result_list = re.split(r"\d+", text)

# 2. Print the resulting list.
print("Resulting list:", result_list)
```

# Task 4: Tokenization Using spaCy

```python
import spacy
import nltk

text = "I'm enjoying the NLP course!"

# 1. Use spaCy with a blank English pipeline (no download required!)
nlp = spacy.blank("en")
doc = nlp(text)

print("--- spaCy Tokenization ---")
# 2. Iterate through the text and print each token on a separate line.
for token in doc:
    print(token.text)

# 3. Explanation: spacy.blank("en") creates a blank language instance containing 
# just the core tokenizer rules, avoiding the need to download external data packages.

print("\n--- NLTK Tokenization (Comparison) ---")
# 5. Tokenize the same text using nltk library and notice the differences.
nltk_tokens = nltk.word_tokenize(text)
print(nltk_tokens)
```
